# 🌊 Tutorial: Análisis de GW150914

Este notebook demuestra cómo cargar datos reales de LIGO y calcular sus propiedades físicas básicas.

**Requisitos:**
- Tener instaladas las dependencias del proyecto (`pip install -e .`)
- Conexión a internet para descargar datos de GWOSC (la primera vez).

```python
import sys
sys.path.insert(0, '../src')

from gw_analysis.io.downloader import download_event_strain, load_hdf5_strain
from gw_analysis.processing.snr_calculator import analyze_signal
from gw_analysis.viz.plots import plot_full_analysis
import matplotlib.pyplot as plt
```

## 1. Descarga de Datos

Descargamos el strain del detector Hanford (H1).

```python
event_id = "GW150914"
detector = "H1"

print(f"Downloading {event_id}...")
path = download_event_strain(event_id, detector)
times, strain = load_hdf5_strain(path)
print(f"Loaded {len(strain)} samples.")
```

## 2. Análisis Cuantitativo

Calculamos SNR aproximado y características del chirp.

```python
metrics = analyze_signal(times, strain)
print("Metrics:", metrics)
```

## 3. Visualización Científica

Generamos un informe completo con Strain, Zoom al Merger y PSD.

```python
plot_full_analysis(times, strain, metrics, event_id)
```
